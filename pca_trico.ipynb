# Principal Component Analysis of Raman Spectra

## Setup

Import required packages

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import brokenaxes as bax

from matplotlib import style
from matplotlib.gridspec import GridSpec
from itertools import cycle
from sklearn.decomposition import PCA

from raman_lib.preprocessing import (BaselineCorrector,
                                     RangeLimiter,
                                     SavGolFilter,
                                     Normalizer)

from raman_lib.misc import load_data

style.use("default")
plt.rcParams['figure.figsize'] = [10, 6]
plt.rcParams['figure.dpi'] = 100
```

Settings

```python
# Path where the data is stored
data_path = r"./data/Trico_Vergleich_Beregnung/Nur_Behandelt_Auswertung"

blank_path = r"./data/Trico_Vergleich_Beregnung/Blank_Auswertung"
# Spectral range/ranges to analyze
# Format: [Start, End, Start2, End2, ...]
# Use 'None' as the first or last value to use the whole range
wn_limits = [420, 480, 580, 650, 1050, 1150, 1250, 1320, 1400, 1500, 2830, 2900]

# Window size for smoothing the spectra
# Higher means stronger smoothing
sg_window = 25
```

```python
blank_data = load_data(blank_path)
blank_data["label"] = blank_data["label"].str.slice(0, 1)

X_blank = blank_data.drop(columns=["label", "file"])

X_blank = np.asarray(X_blank)

y_blank = np.asarray(blank_data.label)
y_blank, y_blank_key = pd.factorize(y_blank, sort=True)

for i, label in enumerate(y_blank_key):
    print(f"{label}: {i}")
```

```python
data = load_data(data_path)


labels = data["label"].str.extract("(.)(\d)")
samples, treatments = labels[0], labels[1]

#data = data[samples.isin(["A", "C", "D", "F"])]

if "file" in data.columns:
    X = data.drop(columns=["label", "file"])
else:
    X = data.drop(columns=[""])
wns = np.asarray(X.columns.astype(float))
X = np.asarray(X)

y, y_key = pd.factorize(data["label"], sort=True)

for i, sample in enumerate(y_key):
    print(f"{sample}: {i}")
```

```python
X_blank = BaselineCorrector().fit_transform(X_blank)
rl = RangeLimiter(lim=wn_limits, reference=wns) 
X_blank = rl.fit_transform(X_blank)
X_blank = SavGolFilter(window=sg_window, limits=rl.lim_).fit_transform(X_blank)
X_blank = Normalizer().fit_transform(X_blank)
```

```python
blank_means = pd.DataFrame(X_blank, 
                           columns=wns)
#np.concatenate([wns[l0:l1] for l0, l1 in zip(rl.lim_[::2], rl.lim_[1::2])]))
blank_means.insert(0, "sample", blank_data["label"])
blank_means = blank_means.groupby("sample").mean()
```

```python
for i in range(len(y_key)):
    plt.plot(wns, X[y==i].mean(axis=0), label=y_key[i])
plt.margins(x=0)
plt.legend()

plt.grid()
plt.xlabel("Wavenumber ($cm^{-1}$)")
plt.ylabel("Intensity (-)")
```

```python
X_bl = BaselineCorrector().fit_transform(X)

for i in range(len(y_key)):
    plt.plot(wns, X_bl[y==i].mean(axis=0), label=y_key[i])
plt.margins(x=0)
plt.legend()

plt.grid()
plt.xlabel("Wavenumber ($cm^{-1}$)")
plt.ylabel("Intensity (-)")
```

```python
rl = RangeLimiter(lim=wn_limits, reference=wns)
X_rl = rl.fit_transform(X_bl)
wns_reduced = np.concatenate([wns[l0:l1] for l0, l1 in zip(rl.lim_[::2], rl.lim_[1::2])])

fig = plt.figure()
ax = bax.brokenaxes(xlims=tuple(zip(rl.lim[::2], rl.lim[1::2])), wspace=0.1)

for i in range(len(y_key)):
    ax.plot(wns_reduced, X_rl[y==i].mean(axis=0), label=y_key[i])

ax.legend()
ax.set_xlabel("Wavenumber ($cm^{-1}$)")
ax.set_ylabel("Intensity (-)")
_ = ax.grid()
```

```python
X_smooth = SavGolFilter(window=sg_window, limits=rl.lim_).fit_transform(X_rl)

fig = plt.figure()
ax = bax.brokenaxes(xlims=tuple(zip(rl.lim[::2], rl.lim[1::2])), wspace=0.1)

for i in range(len(y_key)):
    ax.plot(wns_reduced, X_smooth[y==i].mean(axis=0), label=y_key[i])

ax.legend()
ax.set_xlabel("Wavenumber ($cm^{-1}$)")
ax.set_ylabel("Intensity (-)")
_ = ax.grid()
```

```python

X_norm = Normalizer().fit_transform(X_smooth)
#X_norm = Normalizer(limits=rl.lim_).fit_transform(X_smooth)

fig = plt.figure()
ax = bax.brokenaxes(xlims=tuple(zip(rl.lim[::2], rl.lim[1::2])), wspace=0.1)

for i in range(len(y_key)):
    ax.plot(wns_reduced, X_norm[y==i].mean(axis=0), label=y_key[i])

ax.legend()
ax.set_xlabel("Wavenumber ($cm^{-1}$)")
ax.set_ylabel("Intensity (-)")
ax.grid()
```

```python
X_diff = pd.DataFrame(X_norm, columns=wns_reduced)
X_diff.insert(0, "sample", samples)
X_diff = X_diff.groupby("sample").apply(lambda x: x - blank_means.loc[x.name], include_groups=False)
X_diff = np.asarray(X_diff)

fig = plt.figure()
ax = bax.brokenaxes(xlims=tuple(zip(rl.lim[::2], rl.lim[1::2])), wspace=0.1)

for i in range(len(y_key)):
    ax.plot(wns_reduced, X_diff[y==i].mean(axis=0), label=y_key[i])

ax.legend()
ax.set_xlabel("Wavenumber ($cm^{-1}$)")
ax.set_ylabel("Intensity (-)")
_ = ax.grid()
```

```python
n_components = 3
```

```python
# Run PCA with selected no. of components
pca = PCA(n_components=n_components)
X_pca = pca.fit_transform(X_norm)
```

```python
# simple score plot for PC1 and PC2
shapes = ["o", "^", "s", "d", "v", "p", "P", "*", "<", ">"]

for i, marker in zip(range(len(y_key)), cycle(shapes)):
    plt.scatter(X_pca[y==i, 0], X_pca[y==i, 1], label = y_key[i], marker=marker, edgecolors="k", linewidths=0.5)

plt.xlabel(f"PC1 ({(pca.explained_variance_ratio_[0] * 100):.2f} %)")
plt.ylabel(f"PC2 ({(pca.explained_variance_ratio_[1] * 100):.2f} %)")
plt.legend()

plt.grid()
```

```python
# Pairwise score plot with density plots on the diagonal
plot_data = pd.DataFrame(X_pca, columns=[f"PC {i+1}" for i in range(n_components)])
plot_data.insert(0, "Class", y_key[y])

grid = sns.PairGrid(plot_data, hue="Class", hue_order=y_key)
grid.map_offdiag(sns.scatterplot,
                 style=plot_data["Class"],
                 style_order=y_key,
                 markers=[s for i, s in zip(y_key, cycle(shapes))])
grid.map_diag(sns.kdeplot, fill=True)
grid.set()
grid.add_legend(title="")
```

```python
fig = plt.figure()
ax = bax.brokenaxes(xlims=tuple(zip(rl.lim[::2], rl.lim[1::2])), wspace=0.1)

for i, component in enumerate(pca.components_):
    ax.plot(wns_reduced, component, label=f"PC{i+1}")

ax.axhline(y=0, linestyle="--", linewidth=1, color="k")

ax.legend()
ax.set_xlabel("Wavenumber ($cm^{-1}$)")
ax.set_ylabel("Loadings (-)")
_ = ax.grid()
```
